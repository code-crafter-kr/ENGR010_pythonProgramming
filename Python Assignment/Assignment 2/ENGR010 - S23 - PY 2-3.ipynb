<b>8(E).<span hidden>ENGR010SP23PY02GR03EX08</span></b> Use a While loop to repeatedly ask a user to input a number of bags of cement. Each time the user submits a value, calculate and print the required number of bags of sand and gravel assuming a 1:2:3 ratio (a bag of cement is 3.0 cubic foot, but bags of sand and gravel are only 0.5 cubic foot. Also print the number of cubic feet of concrete this will make, using the information in the Engineering Assignmnent 1. End the loop when the user enters 0. Remember to include units and to round to relevant precision. Test your loop with 3 different values.

```python
while True:
    cement_bags = int(input(print("Enter a number of bags of cement:", end = "")))
    print(cement_bags)
    if cement_bags == 0:
        break
    sand_bags = 6 * 2 * cement_bags
    gravel_bags = 6 * 3 * cement_bags
    concrete_volume = cement_bags * 4.5 * 3 # 1cubic foot cement + 2cubic foot sand + 3cubic foot gravel = 4.5cubic foot cement

    print("cement bags = %d" %cement_bags)
    print("sand bags = %d" %sand_bags)
    print("gravel bags = %d" %gravel_bags)
    print("Total concrete = %.2fft3" %concrete_volume)
```

<b>9(E).<span hidden>ENGR010SP23PY02GR03EX09</span></b> Use a For loop to count the number of vowels in the word "harmonica". Do not manually count.

```python
cnt = 0
vowels = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']
for i in "harmonica":
    if i in vowels:
        cnt += 1
print(cnt)
```

<b>10(M).<span hidden>ENGR010SP23PY02GR03EX10</span></b> The Fibonacci numbers follow the sequence 1, 1, 2, 3, 5 .... where each number is the sum of the two previous numbers in the sequence. Use a For loop to calculate and print out the first 10 numbers in the Fibonacci sequence.

```python
box0 = 1 
box1 = 0
box2 = 0
for i in range (10):
    if i == 0:
        print(box0)
        box1 = box0
    else:
        print(box1)
        box2 = box1 + box0
        box0 = box1
        box1 = box2
```
