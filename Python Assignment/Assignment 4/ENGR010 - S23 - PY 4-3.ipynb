<b>1(E).<span hidden>ENGR010SP23PY04GR03EX01</span></b> Define a function with three parameters that can add, subtract, multiply, or divide two numbers. The first parameter is a string variable that indicates the operation to be performed, such as "add" or "subtract". The second and third parameters are the two numbers to be operated upon. Return the result using a return statement, and then print the result. Test your function on the following operations:

2 + 3<br>
4 - 2<br>
3 * 2<br>
3 / 2<br>
3 ** 2<br>
3 % 2<br>

```python
def calculate(operation, num1, num2):
    if operation == "add":
        result = num1 + num2
    elif operation == "subtract":
        result = num1 - num2
    elif operation == "multiply":
        result = num1 * num2
    elif operation == "divide":
        result = num1 / num2
    elif operation == "power":
        result = num1 ** num2
    elif operation == "remainder":
        result = num1 % num2
    else:
        return "Error"
    return result

print(calculate("add", 2, 3))
print(calculate("subtract", 4, 2))
print(calculate("multiply", 3, 2))
print(calculate("divide", 3, 2))
print(calculate("power", 3, 2))
print(calculate("remainder", 3, 2))
```

<b>2(M).<span hidden>ENGR010SP23PY04GR03EX02</span></b> Define a function called <mark style="color:#EEC83A;background:none">list_analyzer</mark> that accepts a single list-type parameter. Inside the function, print the number of elements, the maximum value, the minimum value, and the average value in the inputted list. In each case, make sure you write text to indicate what you are printing, such as "The maximum value is: ".  Test your function on these lists:

\[1, 2, 3, 4, 5\]<br>
\[2, 3, 1, 5, 4\]<br>

<b>3(E).<span hidden>ENGR010SP23PY04GR03EX03</span></b> Define a function called <mark style="color:#EEC83A;background:none">vowel_finder</mark> that accepts a single string-type parameter. This time, however, print the number of vowels first as indicated in the previous problem. After that, print "They are: " followed by the list of vowels in the string, separated by commas, all on one row. Call <mark style="color:#EEC83A;background:none">vowel_finder</mark> with the word "Shub-Niggurath" as input.

```python
def vowel_finder(word):
    vowel = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']
    result = []
    for i in word:
        if i in vowel:
            result.append(i)
    print(f"Number of vowels: {len(result)}")
    print("They are: ", result)

vowel_finder("Shub-Niggurath")
```

<b>5(M).<span hidden>ENGR010SP23PY04GR03EX05</span></b> Define a function that accepts a string-type parameter. The function should read in content of the file with the name designated by the inputted string parameter, then return the number of vowels in the text of that file. Call your function with the string "input_data.txt" as input.

```python
def read_content(file):
    vowel = ['a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U']
    result = 0
    with open(file, 'r') as infile:
        content = infile.read()
        for i in content:
            if i in vowel:
                result += 1
    
    return result

read_content('input_data.txt')
```

<b>6(H).<span hidden>ENGR010SP23PY04GR03EX06</span></b> Define a function that accepts a string-type parameter. In the function, begin by creating an empty list called <mark style="color:#EE6B3A;background:none">file_text</mark>. Read the contents of the file <b>one line at a time</b>. Remember to strip off the carriage return at the end of each line. Use the <mark style="color:green;background:none">split()</mark> command to create a list from the words in the line. Append this list to <mark style="color:#EE6B3A;background:none">file_text</mark>. Do this for every line in the file. You now have a list (<mark style="color:#EE6B3A;background:none">file_text</mark>) that contains all the lines from the file, each as a separate list. Print the text from <mark style="color:#EE6B3A;background:none">file_text</mark>, making sure that you put a space between each word, so that your printing looks identical to the original file.

```python
def read_print(file):
    file_text = []
    with open(file, 'r') as infile:
        for line in infile:
            line = line.rstrip()
            words = line.split(" ")
            file_text.append(words)

    for content in file_text:
        print(" ".join(content))

read_print("input_data.txt")
```

<b>9(M).<span hidden>ENGR010SP23PY04GR03EX09</span></b> Read in data from <mark style="color:purple;background:none">numbers.csv</mark>, then calculate and print the following:
<ul>
    <li>The overall maximum, minimum and average values in the file</li>
    <li>The index numbers for the lines with the maximum and minimum values, along with the values</li>
    <li>The index numbers for the lines having the highest and lowest average values, along with the values</li>
</ul>

```python
my_data = []
with open("numbers.csv", "r") as infile:
    for line in infile:
        line.rstrip()
        words = line.split(",")
        words = list(map(float, words))
        my_data.append(words)

overall_max = -10000000000000
overall_min = 100000000000000
lst_avg = []
lst_max = []
lst_min = []

total = 0 # for the overall average
elements = 0 # for the overall average

for content in my_data:
    total += sum(content)
    elements += len(content)
    each_avg = sum(content) / len(content)
    lst_avg.append(each_avg)
    lst_max.append(max(content))
    lst_min.append(min(content))


overall_avg = total / elements
print("The overall avg = %.2f" %overall_avg)
print("The overall max = %.2f" %max(lst_max))
print("The overall min = %.2f" %min(lst_min))
print("The highest value is in line %d with value %.2f" %(lst_max.index(max(lst_max)) + 1, max(lst_max)))
print("The lowest value is in line %d with value %.2f" %(lst_min.index(min(lst_min)) + 1, min(lst_min)))
print("The highest average value is in line %d with value %.2f" %(lst_avg.index(max(lst_avg)) + 1, max(lst_avg)))
print("The lowest average value is in line %d with value %.2f" %(lst_avg.index(min(lst_avg)) + 1, min(lst_avg)))
```

<b>10(M).<span hidden>ENGR010SP23PY04GR03EX10</span></b> Define a function that accepts two string-type parameters. The function will read in data from a file with the name provided by the first parameter, count how many instances of the string provided by the second parameter occurs in that content, and then print out the final count. Call your function using "tale.txt" and "best" as the first and second inputs, respectively.

<b>Reminder:</b> Do not include embedded occurences where it is part of another word, such as "bestow".

```python
def read_count(file, word):
    total = 0
    with open(file, 'r') as infile:
        for line in infile:
            words = line.split()
            for i in words:
                if i == word:
                    total += 1
    return total

print(read_count('tale.txt', 'best'))
```
