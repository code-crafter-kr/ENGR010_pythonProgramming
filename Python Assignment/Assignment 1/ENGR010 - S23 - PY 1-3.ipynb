<b>5(H).<span hidden>ENGR010SP23PY01GR03EX05</span></b> Use complex multiplication to show that the square root of 2j is 1+j and the square root of -2j is 1-j.

```python
import math, sys

var1 = 2j**(1/2)
print(var1)
var2 = (-2j)**(1/2)
print(var2)
#Python floating point representation limits
#Machine Epsilon can be used to solve the floating point problem and prove that it is the same.
print("square root of 2j is ({0} + {1}j)".format(var1.real - sys.float_info.epsilon, var1.imag - sys.float_info.epsilon))
print("square root of -2j is ({0} + {1}j)".format(var2.real - sys.float_info.epsilon, var2.imag + sys.float_info.epsilon))
```

Create a table showing the yearly growth of <mark style="color:#EE6B3A;background:none">A</mark> over a 20-year period. Right justify all values and be sure to create column headings indicating the years and amounts. Use initial amount <mark style="color:#EE6B3A;background:none">A</mark> of 200 and annual return <mark style="color:#EE6B3A;background:none">p</mark> of 6.

```python
for n in range (1,21):
    print("year:" + "{0:>2}".format(n), end = " ")
    print("amount:" + "{0:>3f}".format(200*((1+6/100)**n)))
```
