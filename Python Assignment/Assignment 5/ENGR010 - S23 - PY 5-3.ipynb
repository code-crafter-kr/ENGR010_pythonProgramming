<b>3(H).<span hidden>ENGR010SP23PY05GR03EX03</span></b> Find the index of the letter "a" at the beginning of the word "almost" inside the sentence "I would rather be doing almost anything else than this homework.".

<b>Hint:</b> You can use the list of characters created in Ex. 02 to start.

```python
# This function can find the other words, too.
def find_words_idx(sentence, word):
    check_sentence = " " + sentence
    for i in range(1, len(check_sentence)):
        if check_sentence[i-1] == " " and check_sentence[i] == word[0]: #if the first letter of the word is a
            temp = check_sentence[i:i+len(word)] # check the next 5 letters after the 'a' == 'almost'
            if temp == word: # if find return index
                return (i-1) + word.index('a') # exact index of the 'a' in the word.
            
sentence = "I would rather be doing almost anything else than this homework."
print(find_words_idx(sentence, "almost"))
```

<b>5(H).<span hidden>ENGR010SP23PY05GR03EX05</span></b> Import any necessary module, if it has not been imported. Write a function that simulates a roll of dice in the game of [<strong>Risk</strong>](https://en.wikipedia.org/wiki/Risk_(game)) with the basis of the attacker can roll up to 3 dice, the defender can roll one or two. The highest roll of the attacker's dice is compared to the highest role of the defender's dice such that the outcome is decided as follows:
<li>If that attacker's die is greater than that defenders die, the defender loses one army.</li>
<li>If the defender's die is greater than or equal to that attacker's die, the attacker loses an army.</li>
<li>If the attacker has rolled two or three dice, and the defender has rolled two, the second highest attacker's die is then compared to the second highest defender's die, with the same rules as for the first die.</li>

The function will accept two parameters: the first parameter is the number of attacker dice (1, 2, or 3) and the second parameter is the number of defender dice (1 or 2). The function will return the outcome of the battle (2 attacker armies lost, 1 attacker and 1 defender army lost, two defender armies lost, 1 defender army lost, or 1 attacker army lost). Run your function several times with a variety of inputs.

```python
import random
def game(attacker, defender):
    attacker_die = sorted([random.randint(1, 6) for _ in range (attacker)])
    defender_die = sorted([random.randint(1, 6) for _ in range (defender)])
    attacker_lose = 0
    defender_lose = 0
    # first_game
    if attacker_die[-1] <= defender_die[-1]:
        attacker_lose += 1
    else:
        defender_lose += 1

    # second_game
    if attacker >= 2 and defender >= 2:
        if attacker_die[-2] <= defender_die[-2]:
            attacker_lose += 1
        else:
            defender_lose += 1

    return (attacker_lose, defender_lose)

a, b = game(3, 2)
print("%d attacker arimes lost, %d defender arimes lost" % (a, b))
a, b = game(3, 1)
print("%d attacker arimes lost, %d defender arimes lost" % (a, b))
a, b = game(1, 1)
print("%d attacker arimes lost, %d defender arimes lost" % (a, b))
a, b = game(2, 2)
print("%d attacker arimes lost, %d defender arimes lost" % (a, b))
```
