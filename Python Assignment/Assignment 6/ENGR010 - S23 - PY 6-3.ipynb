```python
# Import necessary modules
import numpy as np
import matplotlib.pyplot as plt

# Create and print an array with 50 elements, evenly spaced from -2.0 to 0.0
x_values = np.linspace(-2.0, 0.0, 50)
print("x_values:", x_values)

# Define the function f(x) = -3 + x^2 * e^(-x)
def f(x):
    return -3 + x**2 * np.exp(-x)

# Compute the function values for the x_values array
y_values = f(x_values)

# Plotting
plt.plot(x_values, y_values, label='f(x) = -3 + x^2 * e^(-x)')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Plot of f(x) = -3 + x^2 * e^(-x)')
plt.legend()
plt.grid(True)
plt.show()
```

Print out the final result. Finally, verify that your solution by calculating and printing the value of the function at the solution point to show that the result is very close to the desired value.

```python
from scipy.optimize import fsolve

def f(x):
    return -3 + x**2 * np.exp(-x)

solution = fsolve(f, -0.2) 
print("Solution:", solution)

# Verify the solution by calculating the value of the function at the solution point
value = f(-1.03)
print(value)

# it is close but, not same
# the correct solution is -1.0099 approximately -1.1 but it shows -1.033
```

<b>10(H).<span hidden>ENGR010SP23PY06GR03EX10</span></b> Import any necessary module, if it has not been imported. 
Calculate and save the x- and y-coordinates of a projectile in a frictionless environment every 0.1 seconds, from the moment of launch (t=0) to t=20 seconds, using equations:
<br><br>\begin{align}x(t)=x_0+vcos(Ɵ)t\end{align}<br>
\begin{align}y(t)=y_0+vsin(Ɵ)t–\frac{g}{2}t^2\end{align}<br>
where:
<li>$g=9.81~m/s^2$, gravitational acceleration constant </li>
<li>$v=25~m/s$, projectile velocity</li>
<li>$Ɵ=32$&#176;, initial launch angle</li>
<li>$x_0=0~m$, initial x-coordinates </li>
<li>$y_0=52~m$, initial y-coordinates</li>
<li>t, time in seconds after launch</li>
    
Plot the projectile coordinates on a graph with x-coordinates on the x-axis and y-coordinates on the y-axis.

Remember to include details for an informative plot (such as legends, titles, axis labels,...).

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
g = 9.81 
v = 25  
theta = 32
x_0 = 0  
y_0 = 52  
theta_rad = np.radians(theta)

# Time array from 0 to 20 seconds with a step of 0.1 seconds
time = np.arange(0, 20.1, 0.1)

# Equations
x = x_0 + v * np.cos(theta_rad) * time
y = y_0 + v * np.sin(theta_rad) * time - (g / 2) * time ** 2

# Plotting
plt.figure()
plt.plot(x, y, label="Projectile path")
plt.title("Projectile motion in a frictionless environment")
plt.xlabel("x-coordinate (m)")
plt.ylabel("y-coordinate (m)")
plt.legend()
plt.grid()
plt.show()
```
