<b>8(H).<span hidden>ENGR010SP23PY03GR03EX08</span></b> Create a new empty list called <mark style="color:#EE6B3A;background:none">average_weights</mark>. Calculate the average weight for each of the samples in the list created in Ex. 14, <mark style="color:#EE6B3A;background:none">trials</mark>, using For loops, but instead of printing them, append each value to <mark style="color:#EE6B3A;background:none">average_weights</mark>. Then print the <mark style="color:#EE6B3A;background:none">average_weights</mark> list. Finally, print the maximum and minimum values in the <mark style="color:#EE6B3A;background:none">average_weights</mark> list using the format "The maximum weight is xxx, which is sample n" where xxx is the average weight and n is the corresponding sample number.

```python
average_weights = []
for i in trials:
    avg = sum(i)/len(i)
    average_weights.append(avg)

maximum = max(average_weights)
minimum = min(average_weights)
idx_max = average_weights.index(maximum)
idx_min = average_weights.index(minimum)

print("The maximun weight is %f, which is sample %d" %(maximum, idx_max))
print("The minimun weight is %f, which is sample %d" %(minimum, idx_min))
```

<b>9(M).<span hidden>ENGR010SP23PY03GR03EX09</span></b> Create the multiplication table up to 10\*10 and store them in a list of lists. Then print the table, with the integers from 1 to 10 as both the first row and first column, and the rest of the table having the correct row\*column value.

<b>Hint:</b> Remember that the option end="" can be used in the <mark style="color:green;background:none">print</mark> function to turn off creation of a new line that is activated by default.

```python

lst = [[0 for _ in range(10)] for _ in range (10)] #10*10 table

lst[0] = [i for i in range(1, 11)] #fist row 1 - 10

for j in range (1,11): #first colum 1 - 10
    lst[j-1][0] = j

for x in range (1,10):
    for y in range (1,10):
        lst[x][y] = lst[0][x] * lst[y][0] #multiplication table

for a in lst: #print
    for b in a:
        print("%3d" %b, end = " ")
    print("")
```

<b>10(H).<span hidden>ENGR010SP23PY03GR03EX10</span></b> Create a simple encoder by having the user input a word and an integer between 1 and 7. Add the integer value to the ASCII value of each letter of the inputted word to create an encoded word. Print the encoded word.

```python
word = input("Enter a word to encode: ")
num = int(input("Enter a number between 1 and 7: "))

encoded_word = ""

for i in word:
    encoded_letter = chr(ord(i) + num)
    encoded_word += encoded_letter

print("Encoded word:", encoded_word)
```
