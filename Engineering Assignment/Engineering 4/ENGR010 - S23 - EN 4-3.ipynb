## Section 1: Basic Functions

<b>1. Simple conversion functions</b>

I will make several functions to convert unit. These functions will be used in the main function and will be basic and easy to use. It will help reduce unnecessary code usage and increase readability. This little piece of function is going to produce a big result.

```python
#  Your function to convert meters to inches
def meters_to_inches(meter):
    return  (meter *  39.3701)
```

```python
#  Your function to convert inches to meters
def inches_to_meters(inches):
    return (inches / 39.3701)
```

```python
values = [0, 1, 1.7]
for i in values:
    print("The vlalue is %.2f, meter to inch is %f inches, inch to meter is %f meters" % (i, meters_to_inches(i), inches_to_meters(i)))
```

<b>2. Four more Conversion Functions</b>

I will write new for functions, the functions's name will be intuitive and simple.

```python
#  Your functions that can convert square/cubic meters to square/cubic inches or
#  square/cubic inches to square/cubic meters
def square_meters_to_square_inches(meters):
    return meters * (39.3701 * 39.3701)
def square_inches_to_square_meters(inches):
    return inches / (39.3701 * 39.3701)
def cubic_meters_to_cubic_inches(meters):
    return meters * (39.3701 * 39.3701 * 39.3701)
def cubic_inches_to_cubic_meters(inches):
    return inches / (39.3701 * 39.3701 * 39.3701)
```

```python
#  Your three test cases for each of the four new functions, confirming the correctness of your code
for i in values:
    print("The value of i is : %.2f" %i)
    print("square meters %.2f to suqare inches = %.2f" %(i,square_meters_to_square_inches(i)))
    print("square inches %.2f to square meters = %f" %(i,square_inches_to_square_meters(i)))
    print("cubic meters %.2f to cubic inches = %.2f" %(i,cubic_meters_to_cubic_inches(i)))
    print("cubic inches %.2f to cubic meters = %f" %(i,cubic_inches_to_cubic_meters(i)))
```

<b>3\. Write a controller function</b>

In now, I have 6 functions and I will make a function that includes all of them. And, the string parameter will be used to determine which function to call.

```python
# Your controller function
def controller(event, values):
    if event == "inches to meters":
        return inches_to_meters(values)
    elif event == "meters to inches":
        return meters_to_inches(values)
    elif event == "square meters to square inches":
        return square_meters_to_square_inches(values)
    elif event == "square inches to square meters":
        return square_inches_to_square_meters(values)
    elif event == "cubic meters to cubic inches":
        return cubic_meters_to_cubic_inches(values)
    elif event == "cubic inches to cubic meters":
        return cubic_inches_to_cubic_meters(values)
    else:
        print("Invalid event, you can only use the following events: inches to meters, meters to inches, sqaure meters to sqaure inches, cubic meters to cubic inches, cubic_inches_to_cubic_meters")
        return None
```

```python
# Your tests to confirm that it works as designed.
print("1 inches to meters: %f" %controller("inches to meters", 1))
print("1 meters to inches: %f" %controller("meters to inches", 1))
print("1 square meters to square inches: %f" %controller("square meters to square inches", 1))
print("1 square inches to square meters: %f" %controller("square inches to square meters", 1))
print("1 cubic meters to cubic inches: %f" %controller("cubic meters to cubic inches", 1))
print("1 cubic_inches_to_cubic_meters: %f" %controller("cubic inches to cubic meters", 1))
```

## Section 2: Convert Lists of Value

```python
# Your controller function
def controller(event, values):
    event_mapping = {
        "inches to meters": inches_to_meters,
        "meters to inches": meters_to_inches,
        "square meters to square inches": square_meters_to_square_inches,
        "square inches to square meters": square_inches_to_square_meters,
        "cubic meters to cubic inches": cubic_meters_to_cubic_inches,
        "cubic inches to cubic meters": cubic_inches_to_cubic_meters,
    }

    if event not in event_mapping:
        print("Invalid event, you can only use the following events: inches to meters, meters to inches, square meters to square inches, square inches to square meters, cubic meters to cubic inches, cubic inches to cubic meters")
        return None

    if isinstance(values, (int, float)):
        return event_mapping[event](values)
        
    elif isinstance(values, list):
        return [event_mapping[event](value) for value in values]

# test
test_lst = [0, 1, 1.75]

for event in ["inches to meters", "meters to inches", "square meters to square inches", "square inches to square meters", "cubic meters to cubic inches", "cubic inches to cubic meters"]:
    print(f"Testing {event} with values {test_lst}:")
    result = controller(event, test_lst)
    print(f"Result: {result}\n")
```
