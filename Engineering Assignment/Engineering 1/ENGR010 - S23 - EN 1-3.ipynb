** Now that you have learned a little about concrete and what it is, you need to calculate how much concrete you will need. Assume your driveway wil be 4" thick, 30' long, and 24' wide. Use the cell below to calculate how much concrete you will need, in cubic feet. Be sure to use variable names that tell you what the variable is when you read the name.

```python
thickness_inch = 4 #this value could be chaged to feet
length_feet = 30 #It can be modified according to your request.
width_feet = 24

inches_to_feet = 1/12

thickness_feet = thickness_inch * inches_to_feet

volume = thickness_feet * length_feet * width_feet

print("The amount of the concrete : %.2f cubic feet" %volume)
```

** Your boss has asked you to make the concrete from its basic ingredients, so you need to estimate how much of each of these ingredients you will need to make the concrete. According to the cement manufacturer, a 94 pound sack of Portland cement is 1.0 cubic feet in volume (dry volume, not to be confused with coverage volume) and makes 4.5 cubic feet of concrete using the standard 1:2:3 volumetric ratio of dry cement:sand:gravel. Use the cell below and any additional cells you would like to create to calculate the amount of each ingredient. To be prudent, add 10% to account for some potential waste, spillage etc. 

```python
# Calculate the volumes of ingredients to make 1 cubic feet of concrete.
# These values will be standard volume.

stand_cement = 1 / 4.5 # It means that to make 1 cubic feet of concrete, the cement wil be used for 1/4.5 cubic feet.
stand_sand = stand_cement * 2 # Using 1:2:3 volumetric ratio
stand_gravel = stand_cement * 3 # Using 1:2:3 volumetric ratio
potential_waste = 1.1

# We need 240 cubic feet of concrete, also consider the potential waste
total_cement = stand_cement * volume * potential_waste
total_sand = stand_sand * volume * potential_waste
total_gravel = stand_gravel * volume * potential_waste

print("The capacity of the materials required is as follows.")
print("The amount of the cement: %.2f cubic feet" % total_cement)
print("The amount of the sand: %.2f cubic feet" % total_sand)
print("The amount of the gravel: %.2f cubic feet" % total_gravel)

print("-------------------------------------------------------")
# If we know the densitiy of other ingredients(sand, gravel), can calulate the weight of ingredients.
print("The weight of the cement: %.2f pounds" % (total_cement * 94))

# Suppose the density and calculate the weight of sand and gravel.
density_gravel = 105 # in lb/ft3
density_sand = 100 # in lb/ft3

print("The weight of the gravel: %.2f pounds" % (total_gravel * density_gravel))
print("The weight of the sand: %.2f pounds" % (total_sand * density_sand))
```

Write a function that returns the concrete volume for any length and width - pass these as parameters to the function. If subdivide the price_materials function, can also return the number of bags of necessary substances and the price for each material.

```python
#The numbers below are the given values. It can be treated as a constant.
stand_cement = 1 / 4.5
stand_sand = stand_cement * 2 
stand_gravel = stand_cement * 3 
potential_waste = 1.1
density_gravel = 105 # in lb/ft3
density_sand = 100 # in lb/ft3
price_cement_bag = 18.67 #in dollars
price_sand_bag = 6.47 #in dollars
price_gravel_bag = 5.37 #in dollars
```

```python
# <function> inch to feet (for thickness)
def Converter_inch_feet(a):
    a = a * (1/12)
    return a

# <function> returns the concrete volume for any length and width
def get_concrete_volume(a, b, c): #30, 24, 4
    c = Converter_inch_feet(c)
    volume = a * b * c
    return volume

# <function> returns the materials volume for volume of concrete
def get_materials_volume(a): # a is volume of concrete
    total_cement = stand_cement * a * potential_waste
    total_sand = stand_sand * a * potential_waste
    total_gravel = stand_gravel * a * potential_waste

    return [total_cement, total_sand, total_gravel]

def price_materials(lst_a):
    weight_cement_pound = lst_a[0] * 94
    weight_sand_pound = lst_a[1] * density_sand
    weight_gravel_pound = lst_a[2] * density_gravel

    total_cement_bags = int(weight_cement_pound / 94 + 1)
    total_sand_bags = int(weight_sand_pound / 50 + 1)
    total_gravel_bags = int(weight_gravel_pound / 50 + 1)

    total_price = total_cement_bags * price_cement_bag + total_sand_bags * price_sand_bag + total_gravel_bags * price_gravel_bag

    return total_price
```

```python
print("Information about the materials needed to build a road with x length.(It is fixed at 4 inches thick and 24 inches wide.)")
print("-------------------------------------------")
for i in range (20, 37, 2):
    concrete_volume = get_concrete_volume(i, 24, 4)
    lst_materials_volume = get_materials_volume(concrete_volume)
    total_price = price_materials(lst_materials_volume)

    print("Information when the length of the road is %d feets" %i)
    print("Volume of concrete required %.1f cubic feet:" %concrete_volume)
    print("The amount of the cement: %.2f cubic feet" % lst_materials_volume[0])
    print("The amount of the sand: %.2f cubic feet" % lst_materials_volume[1])
    print("The amount of the gravel: %.2f cubic feet" % lst_materials_volume[2])    

    print("The total price: $%.2f" %total_price)
    print("-------------------------------------------")
```
