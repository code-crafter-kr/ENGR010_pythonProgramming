## Section 1: Data Processing

<b>Step 1: Reading in the data</b>

Put the rod value according to the first number of the list according to the machine number and the second number of the list according to the month. At this time, load_data declares initialization to zero.

```python
import csv

# Initialize the rod_data data structure
rod_data = [[[0 for _ in range(20)] for _ in range(3)] for _ in range(5)] # newsted for loop for initialization

# Read the data from machine.csv and store it in the rod_data data structure
with open('machine.csv', 'r') as file:
    csv_reader = csv.reader(file)
    # Skip the first row (header)
    next(csv_reader)
    n = 0
    for row in csv_reader:
        #rod number will 0 ~ 19 index
        x = n % 20
        n += 1
        machine_num = int(row[0])
        month_num = ['January', 'February', 'March'].index(row[1])
        rod_length = float(row[3])
        rod_data[machine_num][month_num][x] = rod_length


# Print some values from rod_data to confirm that it was read correctly
print(rod_data[1][0][4])  # should print 91.4876798263676
print(rod_data[2][2][19])  # should print 93.2869645905594
print(rod_data[0][0][0]) # should print 91.3779668776973

```

## Section 3: Data Simulation (Adv.)

Write a function that meets the requirements first. and create the new file. lastly, print out to check it.

```python
def create_data(num_machines, operators, months, num_measurements, means, stds, filename):
    with open(filename, 'w', newline='') as file: # firstly read the file
        # Create a CSV writer object
        writer = csv.writer(file)
        # Write the header row
        writer.writerow(['machine', 'month', 'operator', 'length'])
        # Loop over the machines
        for machine in range(num_machines):
            operator = operators[machine]
            for month in months: # Loop over the months
                measurements = np.random.normal(means, stds, size=num_measurements) # make the random measurements
            # Write each measurement to the file
                for measurement in measurements:
                    writer.writerow([machine, month, operator, measurement])

# Generate the simulated data
create_data(num_machines=5, operators=['A', 'B', 'C', 'D', 'E'],
            months=['January', 'February', 'March'], num_measurements=12,
            means=91.44, stds=2.45, filename='machine_simulated.csv')



# Read in the file and print out the first and last 5 lines
print("machine data")
with open('machine.csv', 'r') as file:
    csv_reader = csv.reader(file)
    header = next(csv_reader)
    data = list(csv_reader)
    print(header)
    for i in range(5):
        print(data[i])
    for i in range(-1, -6, -1):
        print(data[i])


print("\nMachine_simulated data")
with open('machine_simulated.csv', 'r') as file:
    csv_reader = csv.reader(file)
    header = next(csv_reader)
    data = list(csv_reader)
    print(header)
    for i in range(5):
        print(data[i])
    for i in range(-1, -6, -1):
        print(data[i])
```
