## Section 1: DNA Sequencing

#### Step 1: Reading in the data

the file has the number on each line. Consider that, firstly, I will read the file and spilit it. And the number is always put on the first word. So, by slicing the the list, I can remove the number. and append it to empty string and return it

```python
Sehyoun_Jang(1)

def DNA_read_function(filename):
    lst = []
    with open(filename, 'r') as file:
        temp = ""
        for line in file:
            a = line.rstrip().split()
            a = a[1:]
            for i in a:
                temp += i
    return temp # whole string

lst = (DNA_read_function("DNA_1.txt"))

print(lst)
```

#### Step 2: Reading in the translation table

After constructing a word by recognizing the alphabet and _, the key and value values are created based on : <br>
All unnecessary spaces and unnecessary characters such as ' are removed.

```python
Sehyoun_Jang(1)
dic = {}
with open("DNA_Amino_Acid_Translation_Table.txt", "r" ) as file2:
    alpha_list = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ_'
    for line in file2:
        a = line.rstrip().split(",")
        for x in a:
            temp_A = ""
            temp_B = ""
            sw = 1
            for char in x:
                if sw == 1:
                    if char in alpha_list:
                        temp_A += char
                    elif char == ":":
                        sw = 2
                if sw == 2:
                    if char in alpha_list:
                        temp_B += char
            if temp_A != "" or temp_B != "":
                dic[temp_A] = temp_B 


print(dic)
```

#### Step 3: Create the Amino Acid Sequence from the DNA sequence

Since the three alphabets produce one result value, the value is obtained by inserting the key value into the dictionary accordingly.

```python
Sehyoun_Jang(1)
DNA1 = DNA_read_function("DNA_1.txt")

def translate_DNA(word):
    result = ""
    for i in range(0, len(word)//3):
        temp = dic[word[i*3:i*3+3].upper()]
        result += temp
    return result

print(translate_DNA(DNA1))
```

## Section 2: Mutated DNA

It interprets both original and mutated files using existing functions. After that, use index numbers to compare characters one by one and add them to the dictionary if they are wrong.

```python
Sehyoun_Jang(2)
def compare_mutated_data(original_data, mutated_data):
    dic = {}
    original = DNA_read_function(original_data)
    original_traslated = translate_DNA(original)
    mutated = DNA_read_function(mutated_data)
    mutated_traslated = translate_DNA(mutated)
    
    for i in range (len(original_traslated)):
        if original_traslated[i] != mutated_traslated[i]:
            loc = i # key
            original_char = original_traslated[i]
            mutated_char = mutated_traslated[i]
            value = mutated_char + " instead of " + original_char
            dic[loc] = value
    return dic

print(compare_mutated_data("DNA_1.txt", "DNA_1_mutated.txt"))
print(compare_mutated_data("DNA_2.txt", "DNA_2_mutated.txt"))
print(compare_mutated_data("DNA_3.txt", "DNA_3_mutated.txt"))
```
